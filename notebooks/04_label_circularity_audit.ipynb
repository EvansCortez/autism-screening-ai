# Reusable label-circularity audit
Read `research/CIRCULARITY.md` before running. This compares score dependence in documented screening datasets; it does not validate diagnoses or transfer an adult model to children.


```python
from pathlib import Path
import sys
ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / 'src/circularity_audit.py').exists())
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))
from src.circularity_audit import run_audit
import pandas as pd
```

```python
reports = {}
for name in ['adult', 'child']:
    reports[name] = run_audit(ROOT / f'research/datasets/{name}.json', ROOT / f'research/circularity/{name}')
pd.DataFrame({name: report['score_dependence'] for name, report in reports.items()}).T
```

```python
for name in reports:
    print(name)
    display(pd.read_csv(ROOT / f'research/circularity/{name}/label_by_score.csv'))
    results = pd.read_csv(ROOT / f'research/circularity/{name}/pooled_metrics.csv')
    display(results.groupby('method')[['accuracy', 'precision', 'recall', 'f1', 'roc_auc']].agg(['mean', 'std']))
```
